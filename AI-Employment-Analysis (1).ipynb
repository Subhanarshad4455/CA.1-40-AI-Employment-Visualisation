The goal of this project is to compare employment across relevant occupations, find trends, and learn more about employment in technology-related industries using data visualization tools.

## 2. Dataset

The Occupational Employment and Wage Statistics (OEWS) dataset for May 2023 from the U.S. Bureau of Labor Statistics (BLS) is the source of employment data used in this research. Employment estimates for a variety of professions in the US are provided by the dataset. It contains data like total employment, profession codes, and occupation titles. Each occupation is identified by the `OCC_CODE` and `OCC_TITLE` columns, and the estimated number of workers in each occupation is displayed by the `TOT_EMP` column. This dataset will be used to investigate and contrast employment in software development, data science, and artificial intelligence-related fields.
