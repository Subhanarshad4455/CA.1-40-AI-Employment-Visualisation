# 5. Visualisation 3 - Employment Across Technology Occupations

```python
# Select detailed technology-related occupations
# Exclude broad occupation groups to avoid overlapping categories
technology_jobs = us_df[
    (us_df["O_GROUP"] == "detailed") &
    (
        us_df["OCC_TITLE"].str.contains(
            "software developer|data scientist|"
            "computer systems analyst|information security analyst|"
            "database architect|web developer|"
            "computer programmer|computer network architect",
            case=False,
            na=False
        )
    )
].copy()

# Remove any duplicate occupation codes
technology_jobs = technology_jobs.drop_duplicates(
    subset=["OCC_CODE"]
)

# Sort occupations by total employment
technology_jobs = technology_jobs.sort_values(
    by="TOT_EMP",
    ascending=True
)

# Create horizontal bar chart
plt.figure(figsize=(11, 7))

plt.barh(
    technology_jobs["OCC_TITLE"],
    technology_jobs["TOT_EMP"],
    color="steelblue"
)

# Add chart title and labels
plt.title("Employment Across Technology Occupations - May 2023")
plt.xlabel("Total Employment")
plt.ylabel("Occupation")

# Display full numbers on the horizontal axis
plt.ticklabel_format(axis="x", style="plain")

# Add grid lines
plt.grid(axis="x", linestyle="--", alpha=0.5)

# Adjust spacing
plt.tight_layout()

# Display chart
plt.show()
```

## Explanation of Visualisation 3

# 6. Visualisation 4 - Employment vs Annual Mean Wage

```python
# Select relevant technology occupations using their occupation codes
# These are the same five occupations used in Visualisation 1

wage_jobs = us_df[
    us_df["OCC_CODE"].isin([
        "15-1252",  # Software Developers
        "15-1211",  # Computer Systems Analysts
        "15-2050",  # Data Scientists
        "15-1212",  # Information Security Analysts
        "15-1243"   # Database Architects
    ])
].copy()

# Convert annual mean wages to numeric values
# Any unavailable or non-numeric values become NaN
wage_jobs["A_MEAN"] = pd.to_numeric(
    wage_jobs["A_MEAN"],
    errors="coerce"
)

# Display occupation names, employment and annual mean wages
wage_jobs[["OCC_TITLE", "TOT_EMP", "A_MEAN"]]
```

```python
# Create a figure for the scatter plot
plt.figure(figsize=(10, 6))

# Plot total employment against annual mean wages
plt.scatter(
    wage_jobs["TOT_EMP"],
    wage_jobs["A_MEAN"],
    s=120,
    color="steelblue"
)

# Add occupation names beside each point
for _, row in wage_jobs.iterrows():
    plt.annotate(
        row["OCC_TITLE"],
        (row["TOT_EMP"], row["A_MEAN"]),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=9
    )

# Add chart title and axis labels
plt.title("Employment vs Annual Mean Wage - May 2023")
plt.xlabel("Total Employment")
plt.ylabel("Annual Mean Wage (USD)")

# Display full numbers on both axes
plt.ticklabel_format(axis="x", style="plain")
plt.ticklabel_format(axis="y", style="plain")

# Add grid lines to improve readability
plt.grid(True, linestyle="--", alpha=0.5)

# Adjust layout and display the chart
plt.tight_layout()
plt.show()
```

## Explanation of Visualisation 4

### Objective

This visualization's goal is to examine the relationship between total employment and yearly mean pay for five particular technology-related jobs in the US in May 2023.

# 8. Visualisation 6 - Employment Trends in Technology Occupations (2021-2023)

# Introduction 

This graphic examines employment trends in a few technology-related jobs in the US between 2021 and 2023. In relation to AI, data science, and software development, the analysis focuses on software developers, data scientists, and information security analysts.

The objective is to determine employment shifts over a three-year period and analyze the evolution of employment in these professions.

The employment statistics between 2021 and 2023 will be compared using a line graph to show any gains or declines.

## Data Sources and Collection

The Occupational Employment and Wage Statistics (OEWS) program of the U.S. Bureau of Labor Statistics (BLS) provided the employment data utilized in this visualization.

The analysis covers three years:

2021: The May 2021 All Data Excel file, which was downloaded from BLS, is used to extract employment statistics.

2022: Employment data for the same occupations are derived from official BLS numbers.

2023: The May 2023 All Data Excel dataset used in the previous visualizations is where employment statistics are taken.

Since only three occupation-level employment estimates are needed for this comparison, the entire 2022 Excel dataset is not loaded into the notebook. Rather, the actual BLS source is cited and the published statistics are directly inserted into a tiny DataFrame.

This method allows the employment numbers from all three years to be compared while minimizing needless data processing.

## Data Preparation

The employment datasets are filtered to only include national, cross-industry employment records before the line chart is created.

Three chosen professions are the subject of the analysis:

1: Software Developers (15-1252)

2: Data Scientists (15-2051)

3: Information Security Analysts (15-1212)

The same occupations are identified over time using the occupation codes. In order to avoid overlapping occupational categories, detailed occupation records are chosen and employment values are transformed into numerical format.

The data will be integrated into a single table with the year, occupation, and total employment after the pertinent employment figures have been extracted.

The line graph and employment changes over time will be computed using this supplied dataset.

```python
# Location of the May 2021 BLS dataset
file_path_2021 = r"D:\4th Year\Data Visualisition\all_data_M_2021.xlsx"

# Read the May 2021 All Data sheet
df_2021 = pd.read_excel(
    file_path_2021,
    sheet_name="All May 2021 data"
)

# Display the dataset size
print("Dataset shape:", df_2021.shape)

# Display column names
print("Column names:")
print(df_2021.columns.tolist())

# Display the first five rows
display(df_2021.head())
```

```python
# Select national and cross-industry records
us_2021 = df_2021[
    (df_2021["AREA"] == 99) &
    (df_2021["NAICS_TITLE"] == "Cross-industry") &
    (df_2021["O_GROUP"] == "detailed")
].copy()

# Select the three technology-related occupations
jobs_2021 = us_2021[
    us_2021["OCC_CODE"].isin([
        "15-1252",  # Software Developers
        "15-2051",  # Data Scientists
        "15-1212"   # Information Security Analysts
    ])
][["OCC_CODE", "OCC_TITLE", "TOT_EMP"]].copy()

# Convert employment figures to numeric values
jobs_2021["TOT_EMP"] = pd.to_numeric(
    jobs_2021["TOT_EMP"],
    errors="coerce"
)

# Display the selected occupations
display(jobs_2021)
```

## Combining Employment Data From 2021, 2022 and 2023

```python
# Prepare 2021 employment data
data_2021 = jobs_2021[["OCC_TITLE", "TOT_EMP"]].copy()
data_2021["Year"] = 2021

# Enter published 2022 employment figures
# 2022 employment estimates from the published BLS occupation tables
data_2022 = pd.DataFrame({
    "OCC_TITLE": [
        "Software Developers",
        "Data Scientists",
        "Information Security Analysts"
    ],
    "TOT_EMP": [1534790, 159630, 163690],
    "Year": [2022, 2022, 2022]
})

# Select the same occupations from the 2023 dataset
data_2023 = us_df[
    (us_df["OCC_CODE"].isin([
        "15-1252",
        "15-2051",
        "15-1212"
    ])) &
    (us_df["O_GROUP"] == "detailed")
][["OCC_TITLE", "TOT_EMP"]].copy()

data_2023["Year"] = 2023

# Combine the three years
employment_trends = pd.concat(
    [data_2021, data_2022, data_2023],
    ignore_index=True
)

# Display the combined data
display(employment_trends)
```

### Line Chart - Employment Trends (2021-2023)

```python
# Create a figure for the line chart
plt.figure(figsize=(11, 6))

# Plot employment trends for each occupation
for occupation in employment_trends["OCC_TITLE"].unique():

    # Select data for the current occupation
    occupation_data = employment_trends[
        employment_trends["OCC_TITLE"] == occupation
    ].sort_values("Year")

    # Draw a line connecting employment values across years
    plt.plot(
        occupation_data["Year"],
        occupation_data["TOT_EMP"],
        marker="o",
        linewidth=2,
        label=occupation
    )

# Add chart title and axis labels
plt.title("Employment Trends in Technology Occupations (2021–2023)")
plt.xlabel("Year")
plt.ylabel("Total Employment")

# Show only the three years on the horizontal axis
plt.xticks([2021, 2022, 2023])

# Display full numbers instead of scientific notation
plt.ticklabel_format(axis="y", style="plain")

# Add grid lines and a legend
plt.grid(True, linestyle="--", alpha=0.5)
plt.legend()

# Adjust spacing and display the chart
plt.tight_layout()
plt.show()
```

## Explanation of Visualisation 6

### Objective

This visualization's goal is to examine employment patterns in three particular technology-related jobs in the US between 2021 and 2023. The purpose of the analysis is to determine if employment rose or fell throughout this time.

### Why a Line Chart?

Since it is helpful for showing changes over time, a line chart was selected. It makes it simpler to spot trends and analyze employment development across occupations by connecting employment statistics from many years.

### What Did I Learn?

1: During 1,364,180 in 2021 to 1,656,880 in 2023, software developers were the most employed group.

2: The number of data scientists employed increased significantly from 105,980 in 2021 to 192,710 in 2023.

3: The number of information security analysts increased from 157,220 in 2021 to 175,350 in 2023.

These results point to an increase in employment in fields including information security, data science, and software development. However, the results should not be regarded as direct indicators of AI employment because these professions are not only tied to AI.

### Percentage Change in Employmnet (2021-2023)

```python
# Arrange employment figures by occupation and year
growth_table = employment_trends.pivot(
    index="OCC_TITLE",
    columns="Year",
    values="TOT_EMP"
)

# Calculate percentage change from 2021 to 2023
growth_table["Growth (%)"] = (
    (growth_table[2023] - growth_table[2021])
    / growth_table[2021]
) * 100

# Display the results rounded to two decimal places
display(growth_table.round(2))
```
