# 3. Visualisation 1 - Employment in Selected Technology Occupations

```python
tech_jobs = us_df[
    us_df["OCC_CODE"].isin([
        "15-1252",
        "15-1211",
        "15-2050",
        "15-1212",
        "15-1243"
    ])
][["OCC_TITLE", "TOT_EMP"]].copy()

tech_jobs
```

```python
plt.figure(figsize=(10, 6))

plt.bar(tech_jobs["OCC_TITLE"], tech_jobs["TOT_EMP"])

plt.title("Employment in Selected Technology Occupations – May 2023")
plt.xlabel("Occupation")
plt.ylabel("Total Employment")
plt.xticks(rotation=45, ha="right")

plt.ticklabel_format(axis="y", style="plain")

plt.tight_layout()
plt.show()
```

## Explanation of Visualisation 1

### Objective

This visualization compares the overall employment levels of a few chosen technology-related jobs in the US in May 2023.

### Why a Bar Chart?

As the vocations are categorical data and the total employment values are numerical data, a bar chart was used. Comparing the employment numbers across various occupations is made simple with a bar chart.

### What Did I Learn?

According to the visualization, with 1,656,880 workers, software developers had the largest employment. There were 498,810 computer systems analysts and 192,710 data scientists. Database architects had the lowest employment, with 59,920, while information security analysts had 175,350. This indicates that in May 2023, software development employment was significantly higher than that of the other chosen technological occupations.


# 4. Visualisation 2 - Employment Share of Selected Technology Occupations

```python
# Create a figure for the pie chart
plt.figure(figsize=(8, 8))

# Display each occupation's share of total employment
plt.pie(
    tech_jobs["TOT_EMP"],
    labels=tech_jobs["OCC_TITLE"],
    autopct="%1.1f%%",
    startangle=90
)

# Add a title to the chart
plt.title("Employment Share of Selected Technology Occupations - May 2023")

# Adjust the layout so labels fit properly
plt.tight_layout()

# Display the pie chart
plt.show()
```

## Explanation of Visualisation 1

This visualization's goal is to display the employment percentage share for five chosen technology-related jobs in the US in May 2023.

### Why a Pie Chart?

As it illustrates how each occupation contributes to the total employment of the five specified occupations, a pie chart was used. Comparing the relative share of each occupation is made simple by the percentage labels.

### What Did I Learn?

At 64.1% of the total employment across these five occupations, software developers made up the greatest share, according to the pie graphic. Data scientists made up 7.5%, and computer systems analysts made up 19.3%. Database architects had the lowest percentage, at 2.3%, while information security analysts made up 6.8%. # 5. Visualisation 3 - Employment Across Technology Occupations

```python
# Select a broader range of technology-related occupations
# from the US-wide, cross-industry dataset
technology_jobs = us_df[
    us_df["OCC_TITLE"].str.contains(
        "software|data scientist|computer systems analyst|"
        "information security|database architect|"
        "web developer|computer programmer|network architect",
        case=False,
        na=False
    )
].copy()

# Remove duplicate occupation codes, if any
technology_jobs = technology_jobs.drop_duplicates(
    subset=["OCC_CODE"]
)

# Sort occupations by employment, from lowest to highest
technology_jobs = technology_jobs.sort_values(
    by="TOT_EMP",
    ascending=True
)

# Create a horizontal bar chart
plt.figure(figsize=(10, 7))

plt.barh(
    technology_jobs["OCC_TITLE"],
    technology_jobs["TOT_EMP"]
)

# Add a title and axis labels
plt.title("Employment Across Technology Occupations - May 2023")
plt.xlabel("Total Employment")
plt.ylabel("Occupation")

# Display full employment numbers on the x-axis
plt.ticklabel_format(axis="x", style="plain")

# Adjust spacing and display the chart
plt.tight_layout()
plt.show()
```

## Explanation of Visualisation 3

### Objective

This visualization's goal is to compare employment in a wider range of technology-related jobs in the US in May 2023.

### Why a Horizontal Bar Chart?

Since it is simpler to read lengthy occupation names when comparing numerical employment values, a horizontal bar chart was selected. It is also possible to determine which occupations have higher or lower employment levels by sorting the occupations by employment.

### What Did I Learn?

The graphic illustrates how employment in technology-related fields differs significantly. Compared to many of the more specialized jobs displayed in the figure, software developers have a high employment rate. Employment numbers are lower for professions including database architects, data scientists, and information security analysts. Although some occupation categories may overlap and shouldn't be viewed as separate groupings when calculating combined totals, the graphic aids in highlighting variations in the size of various occupations.



# 6. Visualisation 4 - Employment vs Annual Mean Wage

```python
# Select relevant technology occupations using their occupation codes
# These are the same five occupations used in Visualisation 1

wage_jobs = us_df[
    us_df["OCC_CODE"].isin([
        "15-1252",  # Software Developers
        "15-1211",  # Computer Systems Analysts
        "15-2050",  # Data Scientists
        "15-1212",  # Information Security Analysts
        "15-1243"   # Database Architects
    ])
].copy()

# Convert annual mean wages to numeric values
# Any unavailable or non-numeric values become NaN
wage_jobs["A_MEAN"] = pd.to_numeric(
    wage_jobs["A_MEAN"],
    errors="coerce"
)

# Display occupation names, employment and annual mean wages
wage_jobs[["OCC_TITLE", "TOT_EMP", "A_MEAN"]]
```

```python
# Create a figure for the scatter plot
plt.figure(figsize=(10, 6))

# Plot total employment against annual mean wages
plt.scatter(
    wage_jobs["TOT_EMP"],
    wage_jobs["A_MEAN"],
    s=120,
    color="steelblue"
)

# Add occupation names beside each point
for _, row in wage_jobs.iterrows():
    plt.annotate(
        row["OCC_TITLE"],
        (row["TOT_EMP"], row["A_MEAN"]),
        xytext=(5, 5),
        textcoords="offset points",
        fontsize=9
    )

# Add chart title and axis labels
plt.title("Employment vs Annual Mean Wage - May 2023")
plt.xlabel("Total Employment")
plt.ylabel("Annual Mean Wage (USD)")

# Display full numbers on both axes
plt.ticklabel_format(axis="x", style="plain")
plt.ticklabel_format(axis="y", style="plain")

# Add grid lines to improve readability
plt.grid(True, linestyle="--", alpha=0.5)

# Adjust layout and display the chart
plt.tight_layout()
plt.show()
```

## Explanation of Visualisation 4

### Objective

This visualization's goal is to examine the relationship between total employment and yearly mean pay for five particular technology-related jobs in the US in May 2023.

### Why a Scatter Plot?

Since it is helpful for comparing two numerical variables, a scatter plot was selected. This graph shows the annual mean wage on the vertical axis and the total employment on the horizontal. It is simpler to determine potential correlations between employment levels and salaries because each point represents a distinct occupation.

### What Did I Learn?

At 1,656,880 jobs and an average yearly salary of 138,110 dollars software developers had the most employment, according to the scatter plot. At 59,920, database architects had the lowest employment rate, but their yearly mean salary was comparatively high at 137,030. With 498,810 workers, computer systems analysts had the lowest yearly mean salary of the five professions, at $110,800.

These results indicate that increasing employment does not always translate into higher annual mean wages. However, the graphic only shows five professions, thus it is unable to draw broad conclusions about the technology industry as a whole



# 7. Visualisation 5 - Distribution of Employment Across Technology Occupations

```python
# Select detailed technology-related occupations
# Exclude broad occupation groups to avoid overlapping categories
hist_jobs = us_df[
    (us_df["O_GROUP"] == "detailed") &
    (us_df["OCC_CODE"].str.startswith("15-"))
].copy()

# Convert employment values into numbers
hist_jobs["TOT_EMP"] = pd.to_numeric(
    hist_jobs["TOT_EMP"],
    errors="coerce"
)

# Remove missing employment values
hist_jobs = hist_jobs.dropna(subset=["TOT_EMP"])

# Display the number of occupations included
print("Number of occupations:", len(hist_jobs))

# Create the histogram
plt.figure(figsize=(10, 6))

plt.hist(
    hist_jobs["TOT_EMP"],
    bins=15,
    edgecolor="black"
)

# Add chart title and axis labels
plt.title("Distribution of Employment in Computer and Mathematical Occupations - May 2023")
plt.xlabel("Total Employment")
plt.ylabel("Number of Occupations")

# Display ordinary numbers instead of scientific notation
plt.ticklabel_format(axis="x", style="plain")

# Add grid lines
plt.grid(axis="y", linestyle="--", alpha=0.5)

# Adjust spacing and display the histogram
plt.tight_layout()
plt.show()
```

## Explanation of Visualisation 5

### Objective

This visualization's goal is to examine the employment distribution of 21 specific computer and mathematical jobs in the US in May 2023.

### Why a Histogram?

Since it is helpful for interpreting the distribution of numerical data, a histogram was used. The total employment is broken down into various ranges in this visualization, and the histogram displays the number of occupations that fall into each range. This is useful in determining whether employment levels are concentrated in specific ranges or distributed equally.

### What Did I Learn?

The employment levels for the majority of computer and mathematical jobs are below about 250,000, according to the histogram. There are only a handful occupations with significantly higher employment levels; one occupation has about 1.66 million workers. This suggests that the distribution is right-skewed, which means that employment is concentrated at lower levels for the majority of occupations while employment figures for a small number of occupations are significantly higher. The results show disparities in the number of jobs in data science, artificial intelligence, and other technology-related fields.
